A cantina quer calcular automaticamente o valor total de todos os itens disponíveis no cardápio.

Os itens e seus respectivos preços estão organizados em duas listas:

itens = ["pizza", "suco", "bolo", "refrigerante"]
precos = [10, 5, 7, 8]

Sua missão é criar um programa que:

* Percorra as listas usando `for`
* Mostre cada item com seu respectivo preço
* Some todos os preços
* Mostre o valor total do cardápio ao final

---

📤 Saída

Item: pizza - Preço: R$ 10

Item: suco - Preço: R$ 5

Item: bolo - Preço: R$ 7

Item: refrigerante - Preço: R$ 8

Valor total do cardápio: R$ 30



```python
itens = ["pizza", "suco", "bolo", "refrigerante"]
precos = [10, 5, 7, 8]
valor_final =0

for i in range(len(itens)):
  print(f"Item: {itens[i]} - Preço R$ {precos[i]}")
  valor_final += precos[i]
print(f"Valor total do cardápio R${valor_final}")
```

# Aula 8
## DESAFIO: FAZENDO UM PEDIDO NO CARDÁPIO DIGITAL

A cantina da escola criou um sistema digital para que os alunos possam escolher vários itens do cardápio.

Os itens e seus preços estão organizados em duas listas:

itens = ["pizza", "suco", "bolo", "refrigerante"]
precos = [10, 5, 7, 8]

O sistema deve permitir que o aluno escolha vários itens até decidir finalizar o pedido digitando **-1**.

Você precisa desenvolver um programa que:

* Mostre o cardápio numerado com os preços
* Permita que o usuário escolha itens pelo número
* Some o valor de cada item escolhido
* Use `while` para permitir múltiplas escolhas
* Encerre o programa quando o usuário digitar **-1**
* Mostre o valor total da compra ao final

---

📥 Entrada

Exemplo de entradas digitadas:

0
2
-1

📤 Saída

----- CARDÁPIO -----

0 - pizza (R$ 10)

1 - suco (R$ 5)

2 - bolo (R$ 7)

3 - refrigerante (R$ 8)

-----------------------
Exemplo:

Escolha o número do item: 0

pizza adicionado! Total parcial: R$ 10

Escolha o número do item: 2

bolo adicionado! Total parcial: R$ 17

Escolha o número do item: -1

Valor total da compra: R$ 17



```python
itens = ["pizza", "suco", "bolo", "refrigerante"]
precos = [10, 5, 7, 8]

print("------CARDÁPIO------")
for i in range(len(itens)):
  print(f"{i} - {itens[i]} - R$ {precos[i]}")

print("--------------------")
valor_final =0
while True:
  escolha = int(input("Digite o número do item que você quer, caso queira sair, digite -1   "))

  if escolha == -1:
    break

  if 0<=escolha<=3:
    valor_final += precos[escolha]
    print(f"Item adicionado! valor parcial: R$ {valor_final}")

print(f"Valor total da compra: R$ {valor_final}")
```

```python
# Cálculo de áreas de formas básicas
import math

def area_quadrado(lado):
    return lado ** 2

def area_retangulo(base, altura):
    return base * altura

def area_triangulo(base, altura):
    return (base * altura) / 2

def area_circulo(raio):
    return math.pi * raio ** 2

def area_trapezio(base_maior, base_menor, altura):
    return ((base_maior + base_menor) * altura) / 2

# Testando
print(f"Área do quadrado (lado 5): {area_quadrado(5)}")
print(f"Área do retângulo (4x6): {area_retangulo(4, 6)}")
print(f"Área do triângulo (base 8, altura 5): {area_triangulo(8, 5)}")
print(f"Área do círculo (raio 3): {area_circulo(3):.2f}")
print(f"Área do trapézio (B=10, b=6, h=4): {area_trapezio(10, 6, 4)}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection

# Configuração estilo moderno
plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(10, 10), facecolor='#0a0a0a')
ax.set_facecolor('#0a0a0a')
ax.set_aspect('equal')
ax.axis('off')

# Cores neon
cores_neon = ['#ff006e', '#fb5607', '#ffbe0b', '#8338ec', '#3a86ff', '#06ffa5']

# Gera várias formas geométricas sobrepostas
np.random.seed(42)
for i in range(60):
    n_lados = np.random.choice([3, 4, 5, 6, 8])
    raio = np.random.uniform(0.5, 4.5)
    angulo_ini = np.random.uniform(0, 2*np.pi)
    rotacao = np.random.uniform(0, 2*np.pi)

    angulos = np.linspace(0, 2*np.pi, n_lados+1) + rotacao
    x = raio * np.cos(angulos) + np.random.uniform(-1, 1)
    y = raio * np.sin(angulos) + np.random.uniform(-1, 1)

    cor = np.random.choice(cores_neon)
    ax.plot(x, y, color=cor, linewidth=np.random.uniform(0.5, 2.5),
            alpha=np.random.uniform(0.3, 0.9))

ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_title("NEON GEOMETRY", color='white', fontsize=24,
             fontweight='bold', pad=20, fontfamily='monospace')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

fig, ax = plt.subplots(figsize=(8, 8))

# Quadrado
quadrado = patches.Rectangle((0, 0), 2, 2, color='skyblue', label='Quadrado')
ax.add_patch(quadrado)

# Círculo
circulo = patches.Circle((5, 1), 1, color='salmon', label='Círculo')
ax.add_patch(circulo)

# Triângulo (polígono)
triangulo = patches.Polygon([[0, 4], [2, 4], [1, 6]], color='lightgreen', label='Triângulo')
ax.add_patch(triangulo)

# Hexágono
import numpy as np
angulos = np.linspace(0, 2*np.pi, 7)
hex_x = 6 + np.cos(angulos)
hex_y = 5 + np.sin(angulos)
hexagono = patches.Polygon(list(zip(hex_x, hex_y)), color='gold', label='Hexágono')
ax.add_patch(hexagono)

# Losango
losango = patches.Polygon([[3, 5], [4, 7], [5, 5], [4, 3]], color='plum', label='Losango')
ax.add_patch(losango)

ax.set_xlim(-1, 9)
ax.set_ylim(-1, 8)
ax.set_aspect('equal')
ax.legend(loc='upper right')
ax.set_title("Formas Geométricas com Matplotlib", fontsize=14)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig = plt.figure(figsize=(14, 5))

# --- Cubo ---
ax1 = fig.add_subplot(131, projection='3d')
r = [-1, 1]
for s, e in [(x, y) for x in r for y in r]:
    ax1.plot3D([s, e], [s, s], [r[0], r[0]], 'b')
    ax1.plot3D([s, e], [s, s], [r[1], r[1]], 'b')
    ax1.plot3D([s, s], [s, e], [r[0], r[0]], 'b')
    ax1.plot3D([s, s], [s, e], [r[1], r[1]], 'b')
    ax1.plot3D([s, s], [r[0], r[0]], [s, e], 'b')
    ax1.plot3D([s, s], [r[1], r[1]], [s, e], 'b')
ax1.set_title("Cubo")

# --- Esfera ---
ax2 = fig.add_subplot(132, projection='3d')
u = np.linspace(0, 2 * np.pi, 30)
v = np.linspace(0, np.pi, 30)
x = np.outer(np.cos(u), np.sin(v))
y = np.outer(np.sin(u), np.sin(v))
z = np.outer(np.ones(np.size(u)), np.cos(v))
ax2.plot_surface(x, y, z, color='red', alpha=0.7)
ax2.set_title("Esfera")

# --- Cone ---
ax3 = fig.add_subplot(133, projection='3d')
theta = np.linspace(0, 2*np.pi, 50)
z = np.linspace(0, 1, 50)
theta_grid, z_grid = np.meshgrid(theta, z)
x = z_grid * np.cos(theta_grid)
y = z_grid * np.sin(theta_grid)
ax3.plot_surface(x, y, z_grid, color='green', alpha=0.7)
ax3.set_title("Cone")

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt

def sierpinski(ax, vertices, depth):
    if depth == 0:
        triangulo = plt.Polygon(vertices, edgecolor='black', facecolor='cyan')
        ax.add_patch(triangulo)
    else:
        # Pontos médios
        p1 = ((vertices[0][0] + vertices[1][0])/2, (vertices[0][1] + vertices[1][1])/2)
        p2 = ((vertices[1][0] + vertices[2][0])/2, (vertices[1][1] + vertices[2][1])/2)
        p3 = ((vertices[0][0] + vertices[2][0])/2, (vertices[0][1] + vertices[2][1])/2)

        sierpinski(ax, [vertices[0], p1, p3], depth - 1)
        sierpinski(ax, [p1, vertices[1], p2], depth - 1)
        sierpinski(ax, [p3, p2, vertices[2]], depth - 1)

fig, ax = plt.subplots(figsize=(8, 7))
vertices = [(0, 0), (1, 0), (0.5, 0.866)]
sierpinski(ax, vertices, 5)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title("Triângulo de Sierpinski (Fractal)", fontsize=14)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(2, 3, figsize=(15, 10))

# Função para polígono regular
def poligono(ax, lados, cor):
    angulos = np.linspace(0, 2*np.pi, lados+1)
    x = np.cos(angulos)
    y = np.sin(angulos)
    ax.fill(x, y, color=cor, alpha=0.7, edgecolor='black', linewidth=2)
    ax.set_aspect('equal')
    ax.axis('off')

poligono(axes[0,0], 3, 'red')      # Triângulo
poligono(axes[0,1], 4, 'blue')     # Quadrado
poligono(axes[0,2], 5, 'green')    # Pentágono
poligono(axes[1,0], 6, 'orange')   # Hexágono
poligono(axes[1,1], 8, 'purple')   # Octógono
poligono(axes[1,2], 12, 'gold')    # Dodecágono

plt.suptitle("Polígonos Regulares com Numpy", fontsize=16)
plt.tight_layout()
plt.show()
```
